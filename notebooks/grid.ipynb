```python
num_modes = (5, 4, 2)
dx = jnp.array([0.3, 0.2, 0.1])
ks = [2 * jnp.pi * jnp.fft.fftfreq(n, d=d) for (n, d) in zip(num_modes, dx)]
k_grid = jnp.stack(jnp.meshgrid(*ks, indexing="ij"), axis=-1)
k_norm = jnp.linalg.norm(k_grid, axis=-1).flatten() ** 2
k_norm
```
