```python
x0 = -10
xf = 10 
num_steps = 1000
num_modes = 250

hilbert_space = qx.spaces.PseudoSpectral(x0, xf, num_steps, num_modes)
T = -0.5 * qx.spaces.PseudoSpectralLaplacian().with_exponentiator(qx.exp.ScaleSquareExponentiator())
V = qx.spaces.PseudoSpectralPotentialEnergy(lambda x: x ** 2)
H = T + V
qx.adapt_operator(H, hilbert_space, 0.01)
```
